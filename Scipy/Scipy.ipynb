```python
#Scipy is an open source computer library used for scientific and technical computing 

from scipy.optimize import minimize 

def f(x):
    return (x-3)**2

res = minimize(f, x0=0) # minimum of a function 
print(res.x)

f = lambda x: (x-3)**2
res = minimize(f, x0=0)
print(res.x)
```

```python
from scipy.optimize import minimize

def f(x):
    return (x**3 * 2**x)

res = minimize(f, x0=2)
print(res.x)

f2 = lambda x: (x**3 * 2**x)
res = minimize(f2, x0=2)
print(res.x)
```

```python
def area_circle(r):
    return (const.pi * (r**2))

def force_gravity(M, m, r):
    return (const.G * M * m)/ (r**2)

print("Area of Circle:", area_circle(5))
print("Gravitational Force:", force_gravity(10, 5, 1))
```

```python
# Integration is a fundamental concept in calculus used to calculate areas under curves, volumes and in solving differential equations.
# quad function computes definite integral of a function with respect to a single variable over given interval,
# points can be  +infinite or - infinite to indicate infinite limits.

from scipy.integrate import quad

def f(x):
    return (3 * x**2 + 1)

res, err = quad(f, 0, 1)
print(res)
print(err)
print()

f2 = lambda x: 3 * x**2 + 1
res, err = quad(f2, 0, 1)
print(res)
print(err)
```

```python
# dblquad function computes double integral of a function with two variables over a rectangular or curved region.

from scipy.integrate import dblquad

f = dblquad(lambda x, y: x * y, 0, 0.5, lambda x: 0, lambda x: 1 - 2*x)
print(f)

def f2(x, y):
    return( x * y)
res = dblquad(f2, 0, 0.5, 
       lambda x: 0, 
       lambda x:1 - 2*x)
print(res)
```

```python
# nquad function computes integrals of functions with multiple variables. It is a flexible method for performing n-dimensional integration over specified limits

from scipy.integrate import nquad

def f(x, y, z):
    return (x * y * z)

I = nquad(f, [[0, 1], [0, 3], [0, 5]])
print(I)

n = lambda x, y, z: x * y * z 
limits = [[0, 1], [0, 3], [0, 5]]
res = nquad(n, limits)
print(res)
```

```python
#cumulative_trapezoid() function computes cumulative integral of discrete data using the trapezoidal rule. 
# Returns the running total of area showing how the integral accumulates step by step.

from scipy.integrate import cumulative_trapezoid

x = np.arange(0, 5)
y = np.arange(0, 5)
s= cumulative_trapezoid(x, y)
print(s)

a = np.arange(0, 3)
b = np.arange(1, 4)
t = cumulative_trapezoid(a, b)
print(t)
```

```python
# simpson() function estimates area under a curve using Simpson’s rule. It works with discrete data points and 
# gives more accurate results than the trapezoidal rule especially for smooth curves.

from scipy.integrate import simpson

x = np.arange(0, 5)
y = np.arange(0, 5)
s= simpson(y, x=x)
print(s)

a = np.arange(0, 3)
b = np.arange(1, 4)
t = simpson(a, x=b)
print(t)
```

```python
# SciPy provides special mathematical functions through scipy.special module. These functions include advanced computations like gamma functions, 
# Bessel functions, error functions, beta functions etc. 

# cbrt : returns the cube root of a number. comb : calculates the number of combinations.
# exp10 : computes 10 raised to the power x. exprel : returns (exp(x) - 1) / x, useful in limiting cases.
# gamma : generalization of factorial: gamma (n + 1) = n! for natural numbers. lambertw : computes the Lambert W function, where W(z) . eW(z) = z
# logsumexp : returns the logarithm of the sum of exponentials useful for numerical stability. perm : calculates the number of permutations
```
