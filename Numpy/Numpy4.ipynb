```python
import numpy as np
```

```python
arr = np.random.randint(0, 5, size = (3, 5))
print(arr)
print()

ar = np.random.randint(1, 10, size = (4, 4))
print(ar)
print()

ars = np.random.randint(3, 15, size = (2, 2, 4))
print(ars)
```

```python
# np.random.normal(loc=0.0, scale=1.0, size=None) is used to generate normal distribution data
# loc(mean) means location moves the distribution left lr right, scale(std) means spread how wide or narrow the distribution is

x = np.random.normal()
print(x)
print()

y = np.random.normal(size = 4); 
print(y)
print()

z = np.random.normal(loc = 10, scale = 2, size = (2, 3))
print(z)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
```

```python
data = np.random.normal(loc=0, scale=1, size=1000)
plt.hist(data, bins=30, edgecolor='red', density=True) # density means dont show counts show probability density instead

x = np.linspace(-4, 4, 200) # start stop and points generation which lead to more curve
pdf = norm.pdf(x, loc=0, scale=1) # norm.pdf gives probability density function of a normal distribution

plt.plot(x, pdf, label='My Histogram')
plt.title('Distribution Histogram')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.grid(True)
plt.legend()
plt.show()
```

```python
# Binomial Distribution is the number of successes in a fixed  number of independent trial of 2 outcomes failure or success
# np.random.binomial(n, p, size=None) n = no of trials, p = percentage of success

x = np.random.binomial(n=10, p=0.5) 
print(x)

y = np.random.binomial(n=12, p=0.6, size=5)
print(y)

m = np.random.binomial(12, 0.6, size=(2, 3))
print(m)
```

```python
from scipy.stats import binom
```

```python
n = 10
p = 0.5
size = 1000
data = np.random.binomial(n, p, size)

# alpha 1.0, 0.0, 0.7 means fully solid, invisible slightly see through
plt.hist(data, bins=np.arange(-0.5, n+1.5, 1), density=True, edgecolor='black', alpha=0.7, label='histogram') 
x = np.arange(0, n+1)
pmf = binom.pmf(x, n, p)

plt.scatter(x, pmf, color='purple', label='Theoretical PMF')
plt.vlines(x, 0, pmf, colors='red', linestyles='dashed')

plt.title("Binomial Distribution (n=10, p=0.5)")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")

plt.legend()
plt.grid()
plt.show()
```

```python
# Poisson distribution models how many times an event occured within a fixed interval when an average occurence rate is known v
# numpy.random.poisson(lam=1.0, size=None)

x = np.random.poisson(lam=3)
print(x)

y = np.random.poisson(lam=5, size=4)
print(y)

z = np.random.poisson(lam=5, size=(3, 3))
print(z)
```

```python
lam = 3
data = np.random.poisson(lam, size=1000)

plt.hist(data, bins=np.arange(-0.5, max(data) + 1.5),  edgecolor='black', density=True)
plt.title(f"Poisson Distribution (λ={lam})")
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.grid(True)
plt.show()
```

```python
# Uniform ditribution is when every value in a given range has an equal probability of occuring 
# np.random.uniform(low=0.0, high=1.0, size=None) low means minimum possible value, high means maximum possible value.

x = np.random.uniform()
print(x)

y = np.random.uniform(size = 3)
print(y)

z = np.random.uniform(10, 20, size = 5)
print(z)
```

```python
# NumPy now recommends using the Generator class for random number generation instead of the legacy numpy.random functions. 
# np.random.default_rng()

rng = np.random.default_rng()
m = rng.uniform(1, 5, size=(2, 3))
print(m)
```

```python
import seaborn as sn
```

```python
low = 0
high = 20
size = 100
data = np.random.uniform(low, high, size)

# kde kernel denisty estimate  smoothed‑out version of the histogram that estimates the underlying distribution. it shows data as a smooth curve
sn.histplot(data, bins=30, kde=False, color='skyblue', edgecolor='pink')
plt.title(f"Uniform Distribution (Range: {low} to {high})")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

```python
# Exponent Distribution is a continous Probability distribution that describe the time between 2 events in poisson process 
# where events occurs independently at a constant average rate np.random.exponential(scale=1.0, size=None)

x = np.random.exponential()
print(x)

y = np.random.exponential(scale=1, size=3)
print(y)
```

```python
s = 2
n = 800
data = np.random.exponential(scale = s, size = n)

sn.histplot(data, bins=30, kde=False, edgecolor='black')
plt.title(f"Exponential Distribution (Scale={s})")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

```python
# The chi-square appears when the squares of standard normal variables are summed up
# np.random.chisquare(df, size=None)

x = np.random.chisquare(df=2)
print(x)

arr = np.random.chisquare(df=2, size=5)
print(arr)
```

```python
from scipy.stats import chi2
```

```python
df = 2
size = 1000

data = np.random.chisquare(df, size)
plt.hist(data, bins=30, density=True, edgecolor='red', alpha=0.7)

x = np.linspace(0, max(data), 200)
pdf = chi2.pdf(x, df)
plt.plot(x, pdf, color='red', label='Theoretical PDF')

plt.title(f"Chi-Square Distribution (df={df})")
plt.xlabel("Value")
plt.ylabel("Density")
plt.legend()
plt.grid(True)
plt.show()
```

```python
array = np.arange(10)
print(array)

r1 = np.mean(array)
print("\nMean: ", r1)

r2 = np.std(array)
print("\nstd: ", r2)

r3 = np.var(array)
print("\nvariance: ", r3)
```

```python
array = np.arange(10)
print(array)

r1 = np.average(array)
print("\nMean: ", r1)

r2 = np.sqrt(np.mean((array - np.mean(array)) ** 2))
print("\nstd: ", r2)

r3 = np.mean((array - np.mean(array)) ** 2)
print("\nvariance: ", r3)
```

```python
x = np.arange(5)
print(x)

r11 = np.mean(x)
r12 = np.average(x)
print("\nMean: ", r11, r12)

r21 = np.std(x)
r22 = np.sqrt(np.mean((x - np.mean(x)) ** 2))
print("\nstd: ", r21, r22)

r31 = np.var(x)
r32 = np.mean((x - np.mean(x)) ** 2)
print("\nvariance: ", r31, r32)
```
