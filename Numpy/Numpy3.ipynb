```python
x = np.array([[1, 2], [4, 5]])
y = np.array([[7, 8], [9, 10]])

print("Square root of X:\n", np.sqrt(x))
print("Square root of Y:\n", np.sqrt(y))
print()

print("Sum of X: ", np.sum(x))
print("Sum of Y: ", np.sum(y))
print("Sum of X and Y (element-wise):\n", x + y)
print()

print("Column-wise sum of X:\n", np.sum(x, axis=0))
print("Row-wise sum of X:\n", np.sum(x, axis=1))
```

```python
A = [[1,2],[4,5]]
B = [[7,8],[9,10]]
rows = len(A)
cols = len(A[0])

C = [[0 for i in range(cols)] for j in range(rows)]
for i in range(rows):
    for j in range(cols):
        C[i][j] = A[i][j] + B[i][j]
print("Addition:\n", C)

D = [[0 for i in range(cols)] for j in range(rows)]
for i in range(rows):
    for j in range(cols):
        D[i][j] = A[i][j] - B[i][j]
print("Subtraction:\n", D)

E = [[0 for i in range(cols)] for j in range(rows)]
for i in range(rows):
    for j in range(cols):
        E[i][j] = A[i][j] / B[i][j]
print("Division:\n", E)
```

```python
a= np.array([2, 6])
b = np.array([3, 10])
c = np.inner(a, b)
print("Inner product of vectors a and b = ", c)

x = np.array([[2, 3, 4], [3, 2, 9]])
y = np.array([[1, 5, 0], [5, 10, 3]])
z = np.inner(x, y)
print("Inner product of matrices x and y = ", z)
```

```python
a = np.array([2, 6])
b = np.array([3, 10])
c = np.outer(a, b)
print("Outer product of vectors a and b = ", c)
```

```python
a = np.array([3, 6, 0])
b = np.array([9, 10, 0])
c = np.cross(a, b)
print("Cross product of vectors a and b: ", c)

x = np.array([[2, 6, 9], [2, 7, 3]])
y = np.array([[7, 5, 6], [3, 12, 3]])
z = np.cross(x, y)
print("Cross product of matrices x and y: ", z)
```

```python
# umpy.linalg.slogdet()  is used to get determinant, it computes the sign and the logarithm of the determinant separately

A = np.array([[50, 29], [30, 44]])
sign, logdet = np.linalg.slogdet(A)
res = sign * np.exp(logdet)
print(res)
```

```python
A = np.array([[1, 2], [3, 4]])
re = np.linalg.det(A)
print(re)

sign, logdet = np.linalg.slogdet(A)
res = sign * np.exp(logdet)
print(res)
```

```python
# numpy.linalg.inv() in the NumPy module is used to compute the inverse matrix in Python.
# f det(A) != 0:
#    A_inv = adj(A) / det(A)
# else:
#    print("Inverse doesn't exist")

A = np.array([[6, 1, 1],
              [4, -2, 5],
              [2, 8, 7]])
c = np.linalg.inv(A)
print(c)
print()

B = np.array([[6, 1, 1, 3],
              [4, -2, 5, 1],
              [2, 8, 7, 6],
              [3, 1, 9, 7]])
d = np.linalg.inv(B)
print(d)
print()
```

```python
# numpy.vdot() returns the dot product of two vectors. If the first vector contains complex numbers, 
# vdot() automatically takes its complex conjugate before performing the multiplication.

a = 2 + 3j
b = 4 + 5j
print(np.vdot(a, b))
```

```python
# numpy.linalg.eig()

m = np.array([[1, 2],[2, 3]])
w, z = np.linalg.eig(m)

print("Printing the Original square array:\n",m)
print()

print("Printing the Eigen values of the given square array:\n", w)
print()

print("Printing Right eigenvectors of the given square array:\n", z)
```

```python
a = np.array([[1, 2, 3],
              [2, 3, 4],
              [4, 5, 6]])

b, c = np.linalg.eig(a)
print("Printing the Original square array:\n", a)
print()

print("Printing the Eigen values of the given square array:\n", b)
print()

print("Printing Right eigenvectors of the given square array:\n", c)
```
