```python
x = np.array([1,2,3,4,5,1,2,1,1,1]) # np.bincount creates a frequency array.
res = np.bincount(x) # np.bincount(x) counts how many times a number appears 
res2 = np.bincount(x).argmax() # print the amount of time the largest argument occured 
print(res)
print(res2)
```

```python
n1 = np.arange(5)
print("1D array:")
print(n1)

n2 = np.arange(10).reshape(2, 5)
print("2D array:")
print(n2)

for a, b in np.nditer([n1, n2]):
    print(f'{a}: {b}')
```

```python
a = np.array([1, 2])
b = np.array([4, 6]) # np.meshgrid(a, b) creates two 2D grids
res = np.array(np.meshgrid(a, b)).T.reshape(-1, 2) # This code generates all combinations (Cartesian product) of the values in a and b
print(res)
```

```python
# np.array_equal is the simplest way to check if two arrays are equal, np.array_equal(arr1, arr2) 

ar1 = np.array([[3, 9], [5, 7]])
ar2 = np.array([[3, 9], [5, 7]])
ar3 = np.array([[2, 9], [5, 7]])

res = np.array_equal(ar1, ar2)
res1 = np.array_equal(ar1, ar3)
res2 = np.array_equal(ar2, ar1)

print(res)
print()
print(res1)
print()
print(res2)
```

```python
ar1 = np.array([[3, 9], [5, 7]])
ar2 = np.array([[3, 9], [5, 7]])
ar3 = np.array([[2, 9], [5, 7]])

def check_arr(a, b):
    if(np.array_equal(a, b)):
        print("Equal")
    else:
        print("Not Equal")

print(check_arr(ar1, ar2))
print()
print(check_arr(ar1, ar3))
print()
print(check_arr(ar2, ar1))
```

```python
# Using == and .all() This method first compares arrays element-wise using ==, then checks if all results are True with .all().

arr1 = np.array([[1, 2], [3, 4]])
arr2 = np.array([[1, 2], [3, 4]])

diff = arr1 == arr2
res = diff.all()
print(res)
```

```python
a = np.array([101, 99, 87])
b = np.array([897, 97, 111])

print("Array a:", a)
print("Array b:", b)
print()

print("a > b:", np.greater(a, b))
print("a < b:", np.less(a, b))
print("a >= b:", np.greater_equal(a, b))
print("a <= b:", np.less_equal(a, b))
```

```python
# Using NumPy's Vectorized Comparison: This method performs a direct element-wise comparison between the array and the given value. 
# It works by generating a Boolean array showing where matches occur, and then verifying whether any match exists.

arr = np.array([[2, 3, 0], [4, 1, 6]])
v = 6
out= np.any(arr == v)
print(out)
```

```python
# np.isnin: checks whether one or more given values appear anywhere in the array by testing each value against all array elements and 
# returning a Boolean result for each value.

arr = np.array([[2, 3, 0], [4, 1, 6]])
vals = [12, 40, 6]
res = {v: np.isin(v, arr) for v in vals}
print(res)
```

```python
# Flattening means converting a 2D matrix into a 1D

matrix = np.array([[10, 20, 30], [40, 50, 60]])
f = matrix.flatten()

print(matrix)
print()
print(f)
```

```python
# order='C' Flatten in row-major order (default). order='F' Flatten in column-major order.
# order='A' Flatten in column-major if memory is Fortran-contiguous, else row-major. order='K' Flatten in the order elements occur in memory.

matrix = np.array([[6, 9], [8, 5], [18, 21]])
f = matrix.flatten()

print(matrix)
print()
print(f)
```

```python
matrix = np.array([[6, 9, 12], [8, 5, 2], [18, 21, 24]])
f = matrix.flatten(order='A')

print(matrix)
print()
print(f)
```

```python
# ravel() tries to return a view of the original array (not a copy). It reads values row-wise and presents them as a 1D array.
# flatten() creates a new copy of the data and converts the array into a row-major 1D array.

a = np.array([[1, 2, 3], [4, 5, 6]])
res = a.ravel()

print(a)
print()

print(res)
flat = a.flatten()
print(flat)
```

```python
# reshape(-1) converts any array into a 1D array. The -1 tells NumPy to automatically compute the required length.
# concatenate() joins all inner arrays (rows) one after another, forming a single continuous 1D array.

a = np.array([[1, 2, 3], [4, 5, 6]])
res = a.reshape(-1)
out = np.concatenate(a)

print(a)
print()
print(res)
print(out)
```
