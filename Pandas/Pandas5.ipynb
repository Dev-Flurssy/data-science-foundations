```python
# isnull(): returns a dataframe or boolean value where true represent missing data (NAN), used to find anf fill missing data 

d = {'First Score': [100, 90, np.nan, 95],
        'Second Score': [30, 45, 56, np.nan],
        'Third Score': [np.nan, 40, 80, 98]}

df = pd.DataFrame(d)
mv = df.isnull()
display(mv)
```

```python
# notnull function that returns a dataframe of boolean value where true means it has a valid value useful for focusing on rows with valid data

d = {'First Score': [100, 90, np.nan, 95],
        'Second Score': [30, 45, 56, np.nan],
        'Third Score': [np.nan, 40, 80, 98]}
df = pd.DataFrame(d)
nmv = df.notnull()
display(nmv)
```

```python
# The interpolate() function fills missing values using interpolation techniques such as the linear method.
   
df = pd.DataFrame({"A": [12, 4, 5, None, 1], 
                   "B": [None, 2, 54, 3, None], 
                   "C": [20, 16, None, 3, 8], 
                   "D": [14, 3, None, None, 6]})  
print(df)
print()

df.interpolate(method='linear', limit_direction='forward')
```

```python
# dropna is used to drop rows or columns with nan

dict = {'First Score': [100, 90, np.nan, 95],
        'Second Score': [30, np.nan, 45, 56],
        'Third Score': [52, 40, 80, 98],
        'Fourth Score': [np.nan, np.nan, np.nan, 65]}
df = pd.DataFrame(dict)
display(df.dropna())
print()

# using how can be used to drop rows 
display(df.dropna(how='all'))
print()

# to drop column with atleast one null values 
display(df.dropna(axis=1))
print()

# to drop row with missing values
display(df.dropna(axis=0))
```

```python
# DataFrame.drop_duplicates(subset=None, keep='first', inplace=False)

data = {
    "Name": ["Alice", "Bob", "Alice", "David"],
    "Age": [25, 30, 25, 40],
    "City": ["NY", "LA", "NY", "Chicago"]
}

df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)
print()

cleaned = df.drop_duplicates()
print(cleaned)
```

```python
data = {
    "Name": ["Alice", "Bob", "Alice", "David"],
    "Age": [25, 30, 25, 40],
    "City": ["NY", "LA", "NY", "Chicago"]
}

df = pd.DataFrame(data)
clean_name = df.drop_duplicates(subset=['Name'])
print(clean_name)
print()

clean_name2 = df.drop_duplicates(keep='last')
print(clean_name2)
print()

remove_all =  df.drop_duplicates(keep=False)
print(remove_all)
print()

df.drop_duplicates(inplace=True)
print(df)
```

```python
# ensuring the column in a data has the corrcet datatype we use astype()

data = {'Name': ['John', 'Alice', 'Bob', 'Eve', 'Charlie'], 
        'Age': [25, 30, 22, 35, 28], 
        'Gender': ['Male', 'Female', 'Male', 'Female', 'Male'], 
        'Salary': [50000, 55000, 40000, 70000, 48000],
       'Join Date': ['2021-01-01', '2020-05-22', '2022-03-15', '2021-07-30', '2020-11-11']}

df = pd.DataFrame(data)
df['Age'] = df['Age'].astype(float)
print(df.dtypes)
print()

df['Join Date'] = pd.to_datetime(df['Join Date']) # pd.to_datetime()converting date stored as a string to date
print(df.dtypes)
print()

df = df.astype({'Age':'float64', 'Salary':'str'})
print(df.dtypes)
```

```python
df = pd.DataFrame({'FirstName': ['Vipul', 'Ashish', 'Milan'],
                            "Gender": ["", "", ""],
                            "Age": [0, 0, 0]})

df['Department'] = np.nan
display(df)

nan_value = float("NaN")
df.replace("", nan_value, inplace=True)
df.dropna(how='all', axis=1, inplace=True)

display(df)
```

```python
# identifying mixed data types 

df = pd.DataFrame( [['tom', 10], ['nick', '15'], ['juli', 14.8]], columns=['Name', 'Age'])
for col in df.columns :
    print(col, ':', pd.api.types.infer_dtype(df[col]))
```

```python
df = pd.DataFrame( [['tom', 10], ['nick', '15'], ['juli', 14.8]], columns=['Name', 'Age'])
df[col] = df[col].astype(int)

for col in df.columns:
    print(col, ':', pd.api.types.infer_dtype(df[col]))
```

```python
data_frame = pd.DataFrame([
    ['tom', 10],
    ['nick', '15'],
    ['juli', 14.8]
], columns=['Name', 'Age'])

for column in data_frame.columns:
    data_frame[column] = pd.to_numeric(data_frame[column], errors='coerce')

for column in data_frame.columns:
    print(pd.api.types.infer_dtype(data_frame[column]))
```
