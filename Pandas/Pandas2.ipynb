```python
import pandas as pd

d1 = {'key': ['K0', 'K1', 'K2', 'K3'],
         'key1': ['K0', 'K1', 'K0', 'K1'],
         'Name':['Jai', 'Princi', 'Gaurav', 'Anuj'],
        'Age':[27, 24, 22, 32],}

d2 = {'key': ['K0', 'K1', 'K2', 'K3'],
         'key1': ['K0', 'K0', 'K0', 'K0'],
         'Address':['Nagpur', 'Kanpur', 'Allahabad', 'Kannuaj'],
        'Qualification':['Btech', 'B.A', 'Bcom', 'B.hons']}

df = pd.DataFrame(d1)
df1 = pd.DataFrame(d2)
print(df, "\n\n", df1)
print()

res = pd.merge(df, df1, on=['key', 'key1'])
print(res)
print()

res2 = pd.merge(df, df1, how='left', on=['key', 'key1'])
print(res2)
print()

res3 = pd.merge(df, df1, how='right', on=['key', 'key1'])
print(res3)
print()

res4 = pd.merge(df, df1, how='outer', on=['key', 'key1'])
print(res4)
print()

res5 = pd.merge(df, df1, how='inner', on=['key', 'key1'])
print(res5)
```

```python
#  QuickSort (kind='quicksort') selects a "pivot" element and partitions the dataset into two halves: one with elements smaller than the pivot 
# and the other with elements greater than the pivot.

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve"],
    "Age": [28, 22, 25, 22, 28],
    "Score": [85, 90, 95, 80, 88]}
df = pd.DataFrame(data)

res = df.sort_values(by= 'Age', kind='quicksort')
print(res)
print()

res2 = df.sort_values(by= 'Age', kind='mergesort') # Divides the dataset into smaller subarrays, sorts them and then merges them back together in sorted order.
print(res2)
print()

res3 = df.sort_values(by= 'Age', kind='heapsort')
print(res3) #  It is another comparison-based sorting algorithm that builds a heap data structure to systematically extract the largest 
# or smallest element and reorder the dataset.
print()
```

```python
# custom sorting logic using the key parameter is useful when we need to sort strings in a specific way such as ignoring case sensitivity.

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve"],
    "Age": [28, 22, 25, 22, 28],
    "Score": [85, 90, 95, 80, 88]}
df = pd.DataFrame(data)

sorted_df = df.sort_values(by='Name', key=lambda col: col.str.lower())
print(sorted_df)
```

```python
# A pivot table is a statistical table that summarizes a substantial table like a big dataset

df = pd.DataFrame({'Product': ['Carrots', 'Broccoli', 'Banana', 'Banana', 'Beans', 'Orange', 'Broccoli', 'Banana'],
                   'Category': ['Vegetable', 'Vegetable', 'Fruit', 'Fruit', 'Vegetable', 'Fruit', 'Vegetable', 'Fruit'],
                   'Quantity': [8, 5, 3, 4, 5, 9, 11, 8],
                   'Amount': [270, 239, 617, 384, 626, 610, 62, 90]})

df
```

```python
pivot = df.pivot_table(index=['Product'], values=['Amount'], aggfunc='sum')
print(pivot)
print()

pivot2 = df.pivot_table(index=['Category'], values=['Amount'], aggfunc='sum')
print(pivot2)
print()

pivot3 = df.pivot_table(index=['Product', 'Category'], values=['Amount'], aggfunc='sum')
print(pivot3)
```

```python
# {'mean', 'min'} will get median, mean and minimum of sales respectively
pivot = df.pivot_table(index=['Category'], values=['Amount'], aggfunc={'median', 'mean', 'min'})
print(pivot)
print()

pivot2 = df.pivot_table(index=['Product'], values=['Amount'], aggfunc={'median', 'mean', 'min'})
print(pivot2)
```
