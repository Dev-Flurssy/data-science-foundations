```python
import json

sample_data = {
    "programs": [
        {
            "id": "001",
            "orchestra": "NYC Philharmonic",
            "season": "2023-24",
            "concerts": [
                {"date": "2023-10-01", "location": "Lincoln Center"},
                {"date": "2023-10-05", "location": "Carnegie Hall"}
            ]
        },
        {
            "id": "002",
            "orchestra": "NYC Philharmonic",
            "season": "2023-24",
            "concerts": [
                {"date": "2023-11-12", "location": "Lincoln Center"}
            ]
        }
    ]
}

with open("raw_nyc.json", "w") as f:
    json.dump(sample_data, f, indent=4)

print("File created successfully!")
```

```python
from pandas import json_normalize

with open("raw_nyc.json") as f:
    d = json.load(f)

nyc = json_normalize(d["programs"])
print(nyc.head())
```
