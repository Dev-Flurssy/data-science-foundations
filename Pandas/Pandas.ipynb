```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
# panda is a one dimensional data array capable of holding data of any type

data = np.array(['a', 'b', 'c', 'd', 'e'])
s = pd.Series(data)
print('Panda Series: \n', s)
```

```python
# pandas dataframe is a 2 dimensional data structure labelled with rows and columns

df = pd.DataFrame()
print(df)
print()

a = ['Geeks', 'For', 'Geeks', 'is', 'portal', 'for', 'Geeks'] 
df2 = pd.DataFrame(a, columns=['Name'])
print(df2)
```

```python
data = {'Names':['Tom', 'nick', 'krish', 'jack'], 'Age':[20, 21, 19, 18]}
df = pd.DataFrame(data)
print(df)
```

```python
data = {'Name':['Jai', 'Princi', 'Gaurav', 'Anuj'],
        'Age':[27, 24, 22, 32],
        'Address':['Delhi', 'Kanpur', 'Allahabad', 'Kannauj'],
        'Qualification':['Msc', 'MA', 'MCA', 'Phd']}
df = pd.DataFrame(data)
print(df[['Name', 'Qualification']])
print(df)
```

```python
# creating a dictionary where each key represent a column and pass it into the dataframe

names = ["Aparna", "Pankaj", "Sudhir", "Garvit"]
degrees = ["MBA", "BCA", "MTech", "MBA"]
scores = [90, 40, 80, 98]



df = pd.DataFrame({'name': names, 'degree': degrees, 'score':scores})
print(df)
```

```python
# the zip() pairs element from lists into tuples creating a row wise data combination 

names = ["Aparna", "Pankaj", "Sudhir", "Geeku"]
values = [11, 22, 33, 44]
df = pd.DataFrame(list(zip(names, values)), columns = ['name', 'value'])
print(df)
```

```python
# data that are initialy defined in row wise can be directly converted into a dataframe 

data = [['Tom', 25], ['Krish', 30], ['Nick', 26], ['Juli', 22]]
df = pd.DataFrame(data, columns=['name', 'age'])
print(df)
```

```python
data = [['Tom', 'Reacher', 25], ['Krish', 'Pete', 30], ['Nick', 'Wilson', 26], ['Juli', 'Williams', 22]]
df = pd.DataFrame(data, columns=['fname', 'lname', 'age'])
print(df)
print()

df['age'] = df['age'].astype(float)
print(df)
```

```python
Name = ['tom', 'krish', 'nick', 'juli']
Age = [25, 30, 26, 22]

df = pd.DataFrame(list(zip(Name, Age)), columns=['Name', 'Age'])
print(df)
print()

df2 = pd.DataFrame({'Name': Name, 'Age': Age})
print(df2)
```

```python
data = [{'name': 'Jake', 'age': 25},
        {'name': 'Martin', 'age': 30}]
df = pd.DataFrame.from_dict(data)
print(df)
```

```python
# from_records() method is used to create a DataFrame from structured data like a list of dictionaries.

data = [{'name': 'Jake', 'age': 25},
        {'name': 'Martin', 'age': 30}]
df = pd.DataFrame.from_records(data)
print(df)
```

```python
# json_normalize() function is used to convert JSON-like or semi-structured data into a flat DataFrame.

data = [{'name': 'Jake', 'age': 25},
        {'name': 'Martin', 'age': 30}]
df = pd.json_normalize(data)
print(df)
```
