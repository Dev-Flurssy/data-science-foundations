```python
# apply function is used to iterate over a df

data = pd.read_csv('nba.csv')
df = pd.DataFrame(data)

def fun(value):
    if(value > 150):
        return 'Yes'
    else:
        return 'No'

df['Chubby'] = df['Weight'].apply(fun)
display(df.head(10))
print()
```

```python
# lambda operator is used to apply log transformation, normalize data and bring it to the range of 0-1 for particular column of data

const = df['Age'].max()
df['New age'] = df['Age'].apply(lambda x: x/const)

display(df.head(10))
print()
```

```python
# Data Norminalization is the process of scaling numeric values so they behave fairly inside a ml model

df = pd.DataFrame([
    [180000, 110, 18.9, 1400],  
    [360000, 905, 23.4, 1800],  
    [230000, 230, 14.0, 1300],  
    [60000, 450, 13.5, 1500]
], columns=['Col A', 'Col B', 'Col C', 'Col D'])

print(df)
print()

df.plot(kind='bar')
plt.show()
```

```python
# maximum absolute scaling rescales each value between -1 and 1 dividing all values by the maximum absolute value in the column

max_scaled = df.copy()

for col in max_scaled.columns:
    max_scaled[col] = max_scaled[col] / (max_scaled[col].abs().max())

print(max_scaled)
print()

max_scaled.plot(kind='bar')
```

```python
# min-max-approach aka normalization rescales the feature to a hard and fast range of [0, 1] by subtracting minimum value of the feature then 
# divide by range 

scaled = df.copy()

for col in scaled.columns:
    scaled[col] = (scaled[col] - scaled[col].min())/ (scaled[col].max() - scaled[col].min())

print(scaled)
print()

scaled.plot(kind='bar')
plt.show()
```

```python
# z score aka standardization change the value in each column so they have a mean of 0 and standard deviation of 1

z_scaled = df.copy()

for col in z_scaled.columns:
    z_scaled[col] = (z_scaled[col] - z_scaled[col].mean())/ z_scaled[col].std()

print(z_scaled)
print()

z_scaled.plot(kind='bar')
plt.show()
```

```python
# groupby is used to split df into groups based on one or more columns 

# Splitting: involves dividing the DataFrame into groups based on some criteria. The groups are defined by unique values in one or more columns.

# Applying: a function is applied to each group independently. You can apply various functions to each group such as:
# Aggregation: Calculate summary statistics (e.g., sum, mean, count) for each group.
# Transformation: Modify the values within each group.
# Filtering: Keep or discard groups based on certain conditions.

# DataFrame.groupby(by=None, axis=0, level=None, as_index=True, sort=True, group_keys=True, observed=False, dropna=True)
```

```python
df = pd.read_csv('nba.csv')
team = df.groupby('Team')
print(team.first().head(5))
print()

groupings = df.groupby(['Team', 'Position'])
print(groupings.first().head(5))
```

```python
agg_data = df.groupby(['Team', 'Position']).agg(
    Total_salary = ('Salary', 'sum'),
    Avg_salary = ('Salary', 'mean'),
    Player_count = ('Name', 'count')
)

print(agg_data)
```

```python
# Transformation functions return an object that is indexed the same as the original group. This is useful when you need to apply operations 
# that maintain the original structure of the data, such as normalization or standardization within groups
# Apply group-specific operations while maintaining the original shape of the dataset.


df = pd.read_csv("https://media.geeksforgeeks.org/wp-content/uploads/nba.csv")

df['Rank within Team'] = df.groupby('Team')['Salary'].transform(lambda x: x.rank(ascending=False))
print(df)
```

```python
filtered = df.groupby('Team').filter(lambda x: x['Salary'].mean() >= 1000000)
print(filtered)
```

```python
# inner join returns a Dataframe with only those rows that have common characteristics.

a = pd.DataFrame()

d = {'id': [1, 2, 10, 12],
     'val1': ['a', 'b', 'c', 'd']}

a = pd.DataFrame(d)
b = pd.DataFrame()

d = {'id': [1, 2, 9, 8],
     'val1': ['p', 'q', 'r', 's']}
b = pd.DataFrame(d)


df = pd.merge(a, b, on='id', how='inner')
df
```

```python
# left join all data from first will be displayed but for second only records with key that can be found in the first will be displayed

a = pd.DataFrame()

d = {'id': [1, 2, 10, 12],
     'val1': ['a', 'b', 'c', 'd']}

a = pd.DataFrame(d)
b = pd.DataFrame()

d = {'id': [1, 2, 9, 8],
     'val1': ['p', 'q', 'r', 's']}
b = pd.DataFrame(d)

df = pd.merge(a, b, on='id', how='left')
df
```

```python
# right join all data from second will be displayed but for first only records with key that can be found in the first will be displayed

a = pd.DataFrame()

d = {'id': [1, 2, 10, 12],
     'val1': ['a', 'b', 'c', 'd']}

a = pd.DataFrame(d)
b = pd.DataFrame()

d = {'id': [1, 2, 9, 8],
     'val1': ['p', 'q', 'r', 's']}
b = pd.DataFrame(d)

df = pd.merge(a, b, on='id', how='right')
df
```

```python
# outer returns data from left and right and matches up rows when possible 

a = pd.DataFrame()

d = {'id': [1, 2, 10, 12],
     'val1': ['a', 'b', 'c', 'd']}

a = pd.DataFrame(d)
b = pd.DataFrame()

d = {'id': [1, 2, 9, 8],
     'val1': ['p', 'q', 'r', 's']}
b = pd.DataFrame(d)

df = pd.merge(a, b, on= 'id', how='outer')
df
```

```python
# Index join To merge the Dataframe on indices pass the left_index and right_index arguments as True 
# i.e. both the Dataframes are merged on an index using default Inner Join.

a = pd.DataFrame()

d = {'id': [1, 2, 10, 12],
     'val1': ['a', 'b', 'c', 'd']}

a = pd.DataFrame(d)
b = pd.DataFrame()

d = {'id': [1, 2, 9, 8],
     'val1': ['p', 'q', 'r', 's']}
b = pd.DataFrame(d)

df = pd.merge(a, b, left_index=True, right_index=True)
df
```
